# Baseline CNN with training-only augmentation

Run `chest_xray_train_val_test_split.ipynb` first, including its duplicate-removal cell. This notebook loads the prepared `chest_xray_cnn_splits/{train,val,test}/{NORMAL,PNEUMONIA}` folders, then applies conservative random rotation, zoom, and translation to **training batches only**, following `augment_training_dataset` in `imbalance.ipynb`. It does not change class counts, and validation and test images are left unaugmented.

## Find prepared data and load splits

```python
from pathlib import Path
import kagglehub
import numpy as np
import tensorflow as tf

SEED = 42
IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 10
DATASET_SLUG = "paultimothymooney/chest-xray-pneumonia"

download_dir = Path(kagglehub.dataset_download(DATASET_SLUG))
candidate_dirs = [
    Path.cwd() / "chest_xray_cnn_splits",
    Path.cwd().parent / "chest_xray_cnn_splits",
    download_dir.parent / "chest_xray_cnn_splits",
]
data_dir = next(
    (path for path in candidate_dirs if all((path / split).is_dir() for split in ("train", "val", "test"))),
    None,
)
if data_dir is None:
    raise FileNotFoundError(
        "Prepared splits not found. Run chest_xray_train_val_test_split.ipynb "
        "through its split and duplicate-removal cells first. Checked: "
        + ", ".join(str(path) for path in candidate_dirs)
    )
print(f"Using prepared data from: {data_dir}")

def load_split(split, shuffle):
    return tf.keras.utils.image_dataset_from_directory(
        data_dir / split,
        image_size=IMAGE_SIZE,
        color_mode="grayscale",
        batch_size=BATCH_SIZE,
        label_mode="int",
        shuffle=shuffle,
        seed=SEED if shuffle else None,
    )

train_ds = load_split("train", shuffle=True)
val_ds = load_split("val", shuffle=False)
test_ds = load_split("test", shuffle=False)
class_names = train_ds.class_names
if class_names != ["NORMAL", "PNEUMONIA"]:
    raise ValueError(f"Expected classes ['NORMAL', 'PNEUMONIA']; got {class_names}")
if val_ds.class_names != class_names or test_ds.class_names != class_names:
    raise ValueError("Train, validation, and test folders must have matching classes.")
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)
test_ds = test_ds.prefetch(AUTOTUNE)
print("Class mapping:", {name: index for index, name in enumerate(class_names)})
```

## Apply training-only augmentation

This mirrors the augmentation utility in `imbalance.ipynb`. Transforms preserve the labels and image shape. Set any factor to `0.0` to disable that transform. Validation and test datasets are not passed through the augmentation function.

```python
rotation_factor = 0.02
zoom_factor = 0.05
translation_factor = 0.02

for name, value, maximum in (
    ("rotation_factor", rotation_factor, 0.5),
    ("zoom_factor", zoom_factor, 1.0),
    ("translation_factor", translation_factor, 1.0),
):
    if not 0 <= value <= maximum:
        raise ValueError(f"{name} must be between 0 and {maximum}.")

augmentation_layers = []
if rotation_factor:
    augmentation_layers.append(
        tf.keras.layers.RandomRotation(rotation_factor, fill_mode="reflect", seed=SEED)
    )
if zoom_factor:
    augmentation_layers.append(
        tf.keras.layers.RandomZoom(
            height_factor=(-zoom_factor, zoom_factor),
            width_factor=(-zoom_factor, zoom_factor),
            fill_mode="reflect",
            seed=SEED + 1,
        )
    )
if translation_factor:
    augmentation_layers.append(
        tf.keras.layers.RandomTranslation(
            height_factor=translation_factor,
            width_factor=translation_factor,
            fill_mode="reflect",
            seed=SEED + 2,
        )
    )

if augmentation_layers:
    augmenter = tf.keras.Sequential(augmentation_layers, name="train_augmenter")

    def augment_batch(images, labels):
        augmented = augmenter(images, training=True)
        augmented.set_shape(images.shape)
        return augmented, labels

    augmented_train_ds = train_ds.map(
        augment_batch, num_parallel_calls=tf.data.AUTOTUNE
    ).prefetch(tf.data.AUTOTUNE)
else:
    augmented_train_ds = train_ds

print(
    "Training augmentation enabled:",
    {"rotation": rotation_factor, "zoom": zoom_factor, "translation": translation_factor},
)
```

## Build and train baseline CNN

```python
tf.keras.utils.set_random_seed(SEED)
num_classes = len(class_names)
model = tf.keras.Sequential(
    [
        tf.keras.layers.Input(shape=(*IMAGE_SIZE, 1)),
        tf.keras.layers.Rescaling(1.0 / 255),
        tf.keras.layers.Conv2D(32, 3, padding="same", activation="relu"),
        tf.keras.layers.MaxPooling2D(),
        tf.keras.layers.Conv2D(64, 3, padding="same", activation="relu"),
        tf.keras.layers.MaxPooling2D(),
        tf.keras.layers.Conv2D(128, 3, padding="same", activation="relu"),
        tf.keras.layers.MaxPooling2D(),
        tf.keras.layers.GlobalAveragePooling2D(),
        tf.keras.layers.Dropout(0.3),
        tf.keras.layers.Dense(64, activation="relu"),
        tf.keras.layers.Dense(num_classes, activation="softmax"),
    ],
    name="baseline_cnn_augmentation",
)
model.compile(
    optimizer=tf.keras.optimizers.Adam(),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
model.summary()
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss", patience=3, restore_best_weights=True
)
history = model.fit(
    augmented_train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=[early_stopping],
)
```

## Evaluate on the held-out test set

```python
from sklearn.metrics import classification_report, confusion_matrix

test_loss, test_accuracy = model.evaluate(test_ds, verbose=0)
print(f"Test loss: {test_loss:.4f}")
print(f"Test accuracy: {test_accuracy:.4f}")
y_true = np.concatenate([labels.numpy().reshape(-1) for _, labels in test_ds])
y_pred = np.argmax(model.predict(test_ds, verbose=0), axis=1)
print("\nClassification report:")
print(classification_report(y_true, y_pred, target_names=class_names, digits=4, zero_division=0))
print("Confusion matrix (rows=true, columns=predicted):")
print(confusion_matrix(y_true, y_pred))
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(history.history["accuracy"], label="train")
axes[0].plot(history.history["val_accuracy"], label="validation")
axes[0].set_title("Accuracy")
axes[0].set_xlabel("Epoch")
axes[0].legend()
axes[1].plot(history.history["loss"], label="train")
axes[1].plot(history.history["val_loss"], label="validation")
axes[1].set_title("Loss")
axes[1].set_xlabel("Epoch")
axes[1].legend()
plt.tight_layout()
plt.show()
```
